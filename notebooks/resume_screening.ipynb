```python
# Stage 1: Data Collection
# Cell 1: Import required libraries

import os
import pandas as pd
from PyPDF2 import PdfReader

print("Libraries imported successfully")
```

```python
# Stage 1: Data Collection
# Cell 2: Locate collected resume PDF files

resume_folder = "../data/resumes"

resume_files = [
    file for file in os.listdir(resume_folder)
    if file.lower().endswith(".pdf")
]

print("Number of resume PDFs:", len(resume_files))
print("\nResume files:")

for file in resume_files:
    print(file)
```

```python
# Stage 1: Data Collection
# Cell 3: Convert PDF resumes into text

resume_texts = []

for file in resume_files:
    file_path = os.path.join(resume_folder, file)
    
    reader = PdfReader(file_path)
    
    text = ""
    for page in reader.pages:
        page_text = page.extract_text()
        if page_text:
            text += page_text + "\n"
    
    resume_texts.append({
        "File_Name": file,
        "Resume_Text": text
    })

resume_data = pd.DataFrame(resume_texts)

print("Number of resumes converted:", len(resume_data))
print("\nResume text conversion completed.")
```

```python
# Stage 1: Data Collection
# Cell 4: Verify extracted resume text

print("First resume file:", resume_data["File_Name"].iloc[0])
print("\nExtracted text preview:\n")
print(resume_data["Resume_Text"].iloc[0][:1000])
```

```python
# Stage 1: Data Collection
# Cell 5: Load the job description

job_description_path = "../data/job_description.txt"

with open(job_description_path, "r", encoding="utf-8") as file:
    job_description = file.read()

print("Job description loaded successfully.")
print("\nJob Description:\n")
print(job_description)
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 6: Import NLTK resources

import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer

# Download required NLTK resources
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("wordnet")
nltk.download("punkt_tab")

print("NLTK resources are ready")
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 7: Convert text to lowercase and remove punctuation/special characters

import re

def clean_text(text):
    # Convert text to lowercase
    text = text.lower()
    
    # Remove punctuation and special characters
    text = re.sub(r"[^a-zA-Z0-9\s]", " ", text)
    
    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()
    
    return text

resume_data["Cleaned_Text"] = resume_data["Resume_Text"].apply(clean_text)

print("Lowercase and special-character removal completed.")
print("\nCleaned text preview:\n")
print(resume_data["Cleaned_Text"].iloc[0][:1000])
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 8: Remove stopwords

stop_words = set(stopwords.words("english"))

def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

resume_data["No_Stopwords_Text"] = resume_data["Cleaned_Text"].apply(remove_stopwords)

print("Stopword removal completed.")
print("\nText after stopword removal:\n")
print(resume_data["No_Stopwords_Text"].iloc[0][:1000])
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 9: Tokenize the preprocessed resume text

def tokenize_text(text):
    return word_tokenize(text)

resume_data["Tokens"] = resume_data["No_Stopwords_Text"].apply(tokenize_text)

print("Tokenization completed.")
print("\nFirst resume tokens:\n")
print(resume_data["Tokens"].iloc[0][:50])
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 10: Lemmatize the tokenized resume text

lemmatizer = WordNetLemmatizer()

def lemmatize_tokens(tokens):
    return [lemmatizer.lemmatize(token) for token in tokens]

resume_data["Lemmatized_Tokens"] = resume_data["Tokens"].apply(lemmatize_tokens)

print("Lemmatization completed.")
print("\nFirst resume lemmatized tokens:\n")
print(resume_data["Lemmatized_Tokens"].iloc[0][:50])
```

```python
# Stage 2: Data Preprocessing (NLP)
# Cell 11: Preprocess the job description

job_description_cleaned = clean_text(job_description)

job_description_no_stopwords = remove_stopwords(job_description_cleaned)

job_description_tokens = tokenize_text(job_description_no_stopwords)

job_description_lemmatized = lemmatize_tokens(job_description_tokens)

print("Job description preprocessing completed.")
print("\nPreprocessed job description:\n")
print(" ".join(job_description_lemmatized))
```

```python
# Stage 3: Feature Extraction
# Cell 12: TF-IDF feature extraction

from sklearn.feature_extraction.text import TfidfVectorizer

# Convert lemmatized resume tokens back into text
resume_data["Processed_Text"] = resume_data["Lemmatized_Tokens"].apply(
    lambda tokens: " ".join(tokens)
)

# Convert lemmatized job description tokens back into text
processed_job_description = " ".join(job_description_lemmatized)

# Create TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Combine resumes and job description for the same TF-IDF vocabulary
all_documents = resume_data["Processed_Text"].tolist() + [processed_job_description]

# Generate TF-IDF feature matrix
tfidf_matrix = tfidf_vectorizer.fit_transform(all_documents)

print("TF-IDF feature extraction completed.")
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
# Stage 3: Feature Extraction
# Cell 13: Bag of Words (BoW) feature extraction

from sklearn.feature_extraction.text import CountVectorizer

# Create BoW vectorizer
bow_vectorizer = CountVectorizer()

# Generate BoW feature matrix
bow_matrix = bow_vectorizer.fit_transform(all_documents)

print("Bag of Words feature extraction completed.")
print("BoW matrix shape:", bow_matrix.shape)
```

```python
# Stage 4: Matching Algorithm
# Cell 14: Extract TF-IDF vectors for matching

# Extract TF-IDF vectors for the 10 resumes
resume_vectors = tfidf_matrix[:-1]

# Extract TF-IDF vector for the job description
job_description_vector = tfidf_matrix[-1]

print("Resume vectors shape:", resume_vectors.shape)
print("Job description vector shape:", job_description_vector.shape)
```

```python
# Stage 4: Matching Algorithm
# Cell 15: Calculate cosine similarity between resumes and job description

from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between each resume and the job description
similarity_scores = cosine_similarity(
    resume_vectors,
    job_description_vector
).flatten()

# Add similarity scores to the resume data
resume_data["Cosine_Similarity"] = similarity_scores

print("Cosine similarity calculation completed.")
print("\nCosine similarity scores:")
print(resume_data[["File_Name", "Cosine_Similarity"]])
```

```python
# Stage 4: Matching Algorithm
# Cell 16: Keyword matching

# Convert the job description tokens into a set of keywords
job_keywords = set(job_description_lemmatized)

# Calculate keyword matches for each resume
keyword_match_counts = []
keyword_match_percentages = []

for tokens in resume_data["Lemmatized_Tokens"]:
    resume_keywords = set(tokens)
    
    matched_keywords = job_keywords.intersection(resume_keywords)
    
    keyword_match_counts.append(len(matched_keywords))
    keyword_match_percentages.append(
        (len(matched_keywords) / len(job_keywords)) * 100
    )

resume_data["Keyword_Matches"] = keyword_match_counts
resume_data["Keyword_Match_Percentage"] = keyword_match_percentages

print("Keyword matching completed.")
print("\nKeyword matching results:")
print(
    resume_data[
        ["File_Name", "Keyword_Matches", "Keyword_Match_Percentage"]
    ]
)
```

```python
# Stage 4: Matching Algorithm
# Cell 17: NLP similarity scoring

# Use the cosine similarity between the preprocessed
# resume text and job description as the NLP similarity score

resume_data["NLP_Similarity_Score"] = resume_data["Cosine_Similarity"]

print("NLP similarity scoring completed.")
print("\nNLP similarity scores:")
print(
    resume_data[
        ["File_Name", "NLP_Similarity_Score"]
    ]
)
```

```python
# Stage 5: Ranking System
# Cell 18: Rank resumes based on similarity score

# Sort resumes from highest to lowest NLP similarity score
ranked_resumes = resume_data.sort_values(
    by="NLP_Similarity_Score",
    ascending=False
).reset_index(drop=True)

# Create ranking numbers
ranked_resumes["Rank"] = ranked_resumes.index + 1

print("Resume ranking completed.")
print("\nRanked resumes:")
print(
    ranked_resumes[
        ["Rank", "File_Name", "NLP_Similarity_Score"]
    ]
)
```

```python
# Stage 5: Ranking System
# Cell 19: Show top matching candidates

# Display the top 5 matching resumes
top_candidates = ranked_resumes.head(5)

print("Top matching candidates:")
print(
    top_candidates[
        ["Rank", "File_Name", "NLP_Similarity_Score"]
    ]
)
```

```python
# Stage 5: Ranking System
# Cell 20: Filter low-score resumes

# Set the minimum similarity score for shortlisted resumes
similarity_threshold = 0.20

# Filter resumes that meet or exceed the threshold
shortlisted_resumes = ranked_resumes[
    ranked_resumes["NLP_Similarity_Score"] >= similarity_threshold
].copy()

print("Low-score resume filtering completed.")
print("\nResumes meeting the similarity threshold:")
print(
    shortlisted_resumes[
        ["Rank", "File_Name", "NLP_Similarity_Score"]
    ]
)
```

```python
# Stage 6: Output
# Cell 21: Add candidate names where available

# Candidate names are not available in the extracted resume text.
# Use "Not Available" instead of treating the job title as the name.

resume_data["Candidate_Name"] = "Not Available"

print("Candidate name field added.")
print("\nCandidate names:")
print(
    resume_data[
        ["File_Name", "Candidate_Name"]
    ]
)
```

```python
# Stage 6: Output
# Cell 22: Display final ranking list with resume match score

# Convert similarity score into percentage
ranked_resumes["Match_Score_Percentage"] = (
    ranked_resumes["NLP_Similarity_Score"] * 100
)

# Add candidate names directly using File_Name
ranked_resumes["Candidate_Name"] = ranked_resumes["File_Name"].map(
    resume_data.set_index("File_Name")["Candidate_Name"]
)

print("Final resume ranking list:")
print(
    ranked_resumes[
        ["Rank", "Candidate_Name", "File_Name", "Match_Score_Percentage"]
    ]
)
```

```python
# Stage 6: Output
# Cell 23: Display shortlisted candidates

# Add candidate names to the shortlisted resumes
shortlisted_resumes["Candidate_Name"] = shortlisted_resumes["File_Name"].map(
    resume_data.set_index("File_Name")["Candidate_Name"]
)

# Convert similarity score into percentage
shortlisted_resumes["Match_Score_Percentage"] = (
    shortlisted_resumes["NLP_Similarity_Score"] * 100
)

print("Shortlisted candidates:")
print(
    shortlisted_resumes[
        ["Rank", "Candidate_Name", "File_Name", "Match_Score_Percentage"]
    ]
)
```
